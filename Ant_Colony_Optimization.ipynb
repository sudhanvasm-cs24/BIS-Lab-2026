```python
import random

# Flow between facilities
flow = [
    [0, 5, 2, 4],
    [5, 0, 3, 0],
    [2, 3, 0, 1],
    [4, 0, 1, 0]
]

# Distance between locations
distance = [
    [0, 2, 3, 1],
    [2, 0, 1, 4],
    [3, 1, 0, 2],
    [1, 4, 2, 0]
]

n = len(flow)

# ACO parameters
ants = 100
iterations = 1000
alpha = 1
beta = 2
evaporation = 0.5
Q = 100

pheromone = [[1 for j in range(n)] for i in range(n)]

best_assignment = None
best_cost = float("inf")


# Calculate total assignment cost
def calculate_cost(assignment):
    cost = 0

    for i in range(n):
        for j in range(n):
            cost += flow[i][j] * distance[assignment[i]][assignment[j]]

    return cost


# Create one assignment
def create_assignment():
    assignment = [-1] * n
    available = list(range(n))

    for facility in range(n):
        weights = []

        for location in available:
            # Heuristic: prefer locations close to already assigned facilities
            heuristic = 1

            for other in range(facility):
                other_location = assignment[other]
                heuristic += flow[facility][other] / (
                    distance[location][other_location] + 1
                )

            # Combine pheromone and heuristic
            weight = (pheromone[facility][location] ** alpha) * (
                heuristic ** beta
            )
            weights.append(weight)

        location = random.choices(available, weights=weights)[0]

        assignment[facility] = location
        available.remove(location)

    return assignment


# Main ACO loop
for iteration in range(iterations):
    assignments = []
    costs = []

    for ant in range(ants):
        assignment = create_assignment()
        cost = calculate_cost(assignment)

        assignments.append(assignment)
        costs.append(cost)

        if cost < best_cost:
            best_cost = cost
            best_assignment = assignment[:]

    # Evaporate pheromones
    for i in range(n):
        for j in range(n):
            pheromone[i][j] *= (1 - evaporation)

    # Deposit pheromones
    for ant in range(ants):
        deposit = Q / costs[ant]

        for facility in range(n):
            location = assignments[ant][facility]
            pheromone[facility][location] += deposit

print("Best assignment:", best_assignment)
print("Minimum cost:", best_cost)
```
