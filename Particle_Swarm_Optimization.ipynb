```python
import random

# Objective function
def fitness(x, y):
    return x * x + y * y


# Number of particles
n = 10

# Number of iterations
iterations = 50

# PSO parameters
w = 0.5
c1 = 2
c2 = 2

# Initialize particles
position = []
velocity = []

for i in range(n):
    x = random.uniform(-10, 10)
    y = random.uniform(-10, 10)

    position.append([x, y])
    velocity.append([0, 0])


# Personal best
pbest = []

for i in range(n):
    pbest.append(position[i][:])


# Find global best
gbest = pbest[0][:]

for i in range(n):
    if fitness(pbest[i][0], pbest[i][1]) < fitness(gbest[0], gbest[1]):
        gbest = pbest[i][:]


# PSO algorithm
for iteration in range(iterations):

    for i in range(n):

        # Random values
        r1 = random.random()
        r2 = random.random()

        # Update velocity
        velocity[i][0] = (
            w * velocity[i][0]
            + c1 * r1 * (pbest[i][0] - position[i][0])
            + c2 * r2 * (gbest[0] - position[i][0])
        )

        velocity[i][1] = (
            w * velocity[i][1]
            + c1 * r1 * (pbest[i][1] - position[i][1])
            + c2 * r2 * (gbest[1] - position[i][1])
        )

        # Update position
        position[i][0] += velocity[i][0]
        position[i][1] += velocity[i][1]

        # Check personal best
        current_value = fitness(position[i][0], position[i][1])
        best_value = fitness(pbest[i][0], pbest[i][1])

        if current_value < best_value:
            pbest[i] = position[i][:]

    # Update global best
    for i in range(n):
        if fitness(pbest[i][0], pbest[i][1]) < fitness(gbest[0], gbest[1]):
            gbest = pbest[i][:]


# Display result
print("Best position:", gbest)
print("Minimum value:", fitness(gbest[0], gbest[1]))
```
