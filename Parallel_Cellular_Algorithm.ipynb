```python
import numpy as np

# ---------------------------------------------------------
# Parallel Cellular Algorithm
# Minimize f(x) = x^2 - 4x + 4
# ---------------------------------------------------------

def fitness(x):
    return (x - 2) ** 2


# Parameters
GRID_SIZE = 100
ITERATIONS = 1000

LOWER_BOUND = -10
UPPER_BOUND = 10

np.random.seed(42)

# Initialize population
grid = np.random.uniform(
    LOWER_BOUND,
    UPPER_BOUND,
    (GRID_SIZE, GRID_SIZE)
)

best_x = None
best_fitness = float('inf')


# Main optimization loop
for iteration in range(ITERATIONS):

    # Evaluate fitness
    fitness_values = fitness(grid)

    # Track global best
    min_index = np.unravel_index(
        np.argmin(fitness_values),
        fitness_values.shape
    )

    current_best_x = grid[min_index]
    current_best_fitness = fitness_values[min_index]

    if current_best_fitness < best_fitness:
        best_fitness = current_best_fitness
        best_x = current_best_x

    # Pad grid to handle boundary cells
    padded = np.pad(
        grid,
        1,
        mode='edge'
    )

    # Extract eight neighboring grids
    neighbors = [
        padded[i:i+GRID_SIZE, j:j+GRID_SIZE]
        for i in range(3)
        for j in range(3)
        if i != 1 or j != 1
    ]

    # Stack neighboring values
    neighbors = np.stack(neighbors)

    # Evaluate neighboring fitness
    neighbor_fitness = fitness(neighbors)

    # Find best neighbor for every cell
    best_indices = np.argmin(
        neighbor_fitness,
        axis=0
    )

    best_neighbors = np.take_along_axis(
        neighbors,
        best_indices[np.newaxis, :, :],
        axis=0
    )[0]

    # Synchronous update
    grid = (grid + best_neighbors) / 2

    # Keep values within bounds
    grid = np.clip(
        grid,
        LOWER_BOUND,
        UPPER_BOUND
    )

    if (iteration + 1) % 10 == 0:
        print(
            f"Iteration {iteration + 1:3d} | "
            f"Best x = {best_x:.6f} | "
            f"Best fitness = {best_fitness:.8f}"
        )


# Final result
print("\n-----------------------------------")
print("Final Best Solution")
print("-----------------------------------")

print(f"Best x       = {best_x:.6f}")
print(f"Best fitness = {best_fitness:.8f}")
```
