```python
import random

# Objective function: minimize x^2
def fitness(x):
    return x * x

# Parameters
n = 30
max_iter = 299
lower = -10
upper = 10

# Initialize random wolf positions
wolves = [random.uniform(lower, upper) for i in range(n)]

for iteration in range(max_iter):

    # Calculate fitness of all wolves
    scores = [fitness(x) for x in wolves]

    # Find Alpha, Beta and Delta wolves
    ranked = sorted(zip(scores, wolves))
    alpha = ranked[0][1]
    beta = ranked[1][1]
    delta = ranked[2][1]

    # Decrease a from 2 to 0
    a = 2 - iteration * (2 / max_iter)

    # Update positions of wolves
    for i in range(n):
        A1 = 2 * a * random.random() - a
        A2 = 2 * a * random.random() - a
        A3 = 2 * a * random.random() - a

        C1 = 2 * random.random()
        C2 = 2 * random.random()
        C3 = 2 * random.random()

        X1 = alpha - A1 * abs(C1 * alpha - wolves[i])
        X2 = beta - A2 * abs(C2 * beta - wolves[i])
        X3 = delta - A3 * abs(C3 * delta - wolves[i])

        wolves[i] = (X1 + X2 + X3) / 3

        # Keep position within limits
        wolves[i] = max(lower, min(upper, wolves[i]))

# Find the best solution
best = min(wolves, key=fitness)

print("Best position:", best)
print("Minimum value:", fitness(best))
```
